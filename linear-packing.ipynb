```python
from pathlib import Path

data = Path("/scratch/project_2008523/era5")
```

```python
grib_single = data / "a4a3366b1043dd7fe0d9d4fdbd512a4e.grib"
grib_pressure = data / "fd13d32f1a58ec2a87c51a51e4e4c682.grib"
```

```python
import re

def atoi(s: str):
    return int(s) if s.isdigit() else s

def natural_keys(s: str) -> list[str | int]:
    '''
    alist.sort(key=natural_keys) sorts in human order
    http://nedbatchelder.com/blog/200712/human_sorting.html
    '''
    return [atoi(c) for c in re.split(r"\d+", s)]
```

```python
import eccodes
```

```python
lines = []

with eccodes.highlevel.FileReader(grib_single) as reader:
    for message in reader:
        name = message.get("shortName")
        dataType = message.get("dataType")
        date = str(message.get("date"))
        year, month, day = date[:4], date[4:6], date[6:]
        decimalScaleFactor = message.get("decimalScaleFactor")
        referenceValue = message.get("referenceValue")
        binaryScaleFactor = message.get("binaryScaleFactor")
        bitsPerValue = message.get("bitsPerValue")

        line = f"{name} [{dataType}] @ {day}.{month}.{year} 00:00 UTC: "

        if bitsPerValue > 0:
            line += f"{name} * {10**decimalScaleFactor} = {referenceValue} + x * 2^{binaryScaleFactor} @ {bitsPerValue}b"
        else:
            line += f"{name} * {10**decimalScaleFactor} = {referenceValue} [const]"

        lines.append(line)

print("\n".join(sorted(lines, key=natural_keys)))
```
