```python
import numpy as np
import pandas as pd
import xarray as xr
```

```python
from data import UncertainDataset, ds_single, ds_pressure
```

```python
from dataclasses import dataclass

@dataclass
class MinimumBounds:
    percentile: list[float]
    mean_bin_edges: list[float]
    mean_bin_counts: list[int]
    mean_bin_spread_bounds: list[list[float]]
```

```python
from tqdm.notebook import tqdm
```

```python
def compute_minimum_bound(
    mean: xr.DataArray,
    spread: xr.DataArray,
    percentile: list[float],
    nbins: int = 100,
) -> MinimumBounds:
    mean, spread = mean.copy(deep=True), spread.copy(deep=True)
    
    mean = mean.values.flatten()
    spread = spread.values.flatten()
    
    amin, amax = np.nanmin(mean), np.nanmax(mean)

    mean_bin_edges = np.nanquantile(mean, np.linspace(0.0, 1.0, nbins+1))

    ibin = np.minimum(np.searchsorted(mean_bin_edges, mean), nbins-1)

    mean_bin_spread_bounds = [np.zeros(nbins) for _ in percentile]

    for i in tqdm(range(nbins), position=1, leave=False):
        ispread = spread[ibin == i]

        if len(ispread) > 0:
            bs = np.nanquantile(ispread, [1-p for p in percentile])
        else:
            bs = [np.nan for _ in percentile]

        for bd, b in zip(mean_bin_spread_bounds, bs):
            bd[i] = b

    mean_bin_counts, _ = np.histogram(mean, bins=mean_bin_edges)

    return MinimumBounds(
        percentile=percentile,
        mean_bin_edges=list(mean_bin_edges),
        mean_bin_counts=list(mean_bin_counts),
        mean_bin_spread_bounds=[list(bs) for bs in mean_bin_spread_bounds],
    )
```

```python
with open("minimum-single.txt", "w") as f:
    for key in tqdm(ds_single.mean, position=0):
        print(f"{key}: ", end="", flush=True, file=f)
        print(f"{compute_minimum_bound(ds_single.mean[key], ds_single.spread[key], [1.0, 0.99, 0.95])!r}", flush=True, file=f)
```

```python
with open("minimum-pressure.txt", "w") as f:
    for key in tqdm(ds_pressure.mean, position=0):
        print(f"{key}:", flush=True, file=f)
        for pressure in ds_pressure.mean[key].pressure_level.values:
            print(f" - {pressure} hPa: ", end="", flush=True, file=f)
            print(f"{compute_minimum_bound(ds_pressure.mean[key].sel(pressure_level=pressure), ds_pressure.spread[key].sel(pressure_level=pressure), [1.0, 0.99, 0.95])!r}", flush=True, file=f)
```
